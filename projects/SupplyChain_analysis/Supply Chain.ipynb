```python
import pandas as pd
import numpy as np
```

Load the raw dataset from the CSV file 

```python
file_path = "C:/Users/vaibh/Desktop/DA Project/Project/DataCoSupplyChainDataset.csv"
df = pd.read_csv(file_path, encoding='latin1')
```

Clean up column names by converting to lowercase and replacing spaces with underscores.

```python
df.columns = df.columns.str.lower().str.replace(' ', '_')
df.rename(columns={'days_for_shipping_(real)': 'days_for_shipping_real',
                   'days_for_shipment_(scheduled)': 'days_for_shipment_scheduled',
                   'late_delivery_risk': 'is_late_delivery'}, inplace=True)
df.rename(columns={'order_date_(dateorders)': 'order_date'}, inplace=True)
```

Convert date columns to the datetime format.

```python
df['order_date'] = pd.to_datetime(df['order_date'])
df['shipping_date_(dateorders)'] = pd.to_datetime(df['shipping_date_(dateorders)'])
```

missing values with 0

```python
df.fillna(0, inplace=True)
```

shipping time variance

```python
df['shipping_variance'] = df['days_for_shipping_real'] - df['days_for_shipment_scheduled']
```

Profit_Margin

```python
df['profit_margin'] = df['benefit_per_order'] / df['sales_per_customer']
```

Export the final cleaned DataFrame to a new CSV file

```python
df.to_csv('C:/Users/vaibh/Desktop/DA Project/Project/cleaned_supply_chain_data.csv', index=False)

print("Data cleaning and preparation complete. 'cleaned_supply_chain_data.csv' is ready for analysis.") 
```
